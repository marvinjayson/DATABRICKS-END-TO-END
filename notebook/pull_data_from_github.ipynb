```python
import pandas as pd

# GitHub raw base URL
BASE_URL = "https://raw.githubusercontent.com/marvinjayson/DATABRICKS-END-TO-END/main/data/"

# Map of Delta table name -> CSV file name
FILES = {
    "customers":                        "customers.csv",
    "geolocation":                      "geolocation.csv",
    "order_items":                      "order_items.csv",
    "order_payments":                   "order_payments.csv",
    "order_reviews":                    "order_reviews.csv",
    "orders":                           "orders.csv",
    "product_category_name_translation": "product_category_name_translation.csv",
    "products":                         "products.csv",
    "sellers":                          "sellers.csv",
}

for table_name, file_name in FILES.items():
    url = BASE_URL + file_name
    print(f"Reading {file_name}...")

    # Read from GitHub via pandas
    pdf = pd.read_csv(url)

    # Convert to Spark DataFrame and write as Delta table
    (
        spark.createDataFrame(pdf)
        .write
        .format("delta")
        .mode("overwrite")
        .option("overwriteSchema", "true")
        .saveAsTable(f"sales.bronze.{table_name}")
    )

    print(f"  -> sales.bronze.{table_name} ({pdf.shape[0]:,} rows x {pdf.shape[1]} cols)")

print("\nAll tables loaded into sales.bronze successfully!")
```
